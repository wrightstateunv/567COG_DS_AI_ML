# Data Embedding Pipeline
This notebook reads the combined chunked dataset, generates embeddings for each text chunk, and saves the result for LLM/Foundry ingestion.

```python
# Section 1: Import libraries
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer

# Section 2: Load chunked dataset
chunked_path = '/Users/mathewthomas/Documents/hobby_projects/AI_ML_Work/567COG_DS_AI_ML/Usecases/A9A_Assessment/data_processing/chunked_data/a9a_assessment_single_chunked_document.csv'
df = pd.read_csv(chunked_path)
print(f"Loaded {len(df)} rows from chunked document.")
```

```python
# Section 3: Generate embeddings for each text_chunk using spaCy
import spacy
import ast

nlp = spacy.load("en_core_web_md")  # Make sure to run: python -m spacy download en_core_web_md

def get_spacy_embedding(text):
    return nlp(text).vector.tolist()  # Ensure output is a list

embeddings = df['text_chunk'].fillna('').apply(get_spacy_embedding)
df['embedding_vector'] = embeddings

# Section 4: Combine embeddings with metadata (embedding_vector as a list column)
embedded_df = df.copy()

# Section 5: Save embedded document
embedded_path = '/Users/mathewthomas/Documents/hobby_projects/AI_ML_Work/567COG_DS_AI_ML/Usecases/A9A_Assessment/data_processing/chunked_data/a9a_assessment_single_embedded_document.csv'
embedded_df.to_csv(embedded_path, index=False)
print(f"Saved embedded document: {embedded_path} (rows: {len(embedded_df)})")
```
