# Data Generator for A9A Assessment

This notebook generates the required CSV files as described in the dataset instructions. Each file will be saved in the `../data/` folder. Unknown columns are filled with relevant names, and `Mission_Title` is included in all files to link them together.

## Instructions

This notebook generates the following CSV files in the `../data/` folder:

1. Cyber_Document_Library
2. Equipment_Issues
3. FFIRs_Friction_Points_Recommendations
4. Hardening_Recommendations
5. Mission_Details
6. Mission_Metrics
7. Missions_by_Location_and_Tasked_System
8. Tactical_Lessons_Learned

Each file includes a `Mission_Title` column to link records across files. Unknown columns are filled with relevant names. Long text columns are simulated with sample text.

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import os

# Output directory
DATA_DIR = '../data/'
os.makedirs(DATA_DIR, exist_ok=True)

# Helper function to generate random mission titles
def generate_mission_titles(n):
    return [f"Mission_{i+1}" for i in range(n)]

# Helper function for random text
def random_text(length=50):
    return ' '.join(np.random.choice(['cyber', 'threat', 'intel', 'operation', 'system', 'network', 'security', 'assessment', 'response', 'incident'], size=length))
```

```python
# 1) Cyber_Document_Library
n_rows = 10
mission_titles = generate_mission_titles(n_rows)
cyber_doc_columns = ['Mission_Title', 'Doc_ID', 'Doc_Type', 'Doc_Date', 'Classification', 'Author', 'Review_Status']
cyber_doc_df = pd.DataFrame({
    'Mission_Title': mission_titles,
    'Doc_ID': [f'DOC{i+1:03d}' for i in range(n_rows)],
    'Doc_Type': np.random.choice(['Report', 'Memo', 'Briefing'], n_rows),
    'Doc_Date': pd.date_range(datetime.today(), periods=n_rows).strftime('%Y-%m-%d'),
    'Classification': np.random.choice(['Unclassified', 'Secret', 'Top Secret'], n_rows),
    'Author': np.random.choice(['Analyst A', 'Analyst B', 'Analyst C'], n_rows),
    'Review_Status': np.random.choice(['Pending', 'Reviewed', 'Rejected'], n_rows)
})
cyber_doc_df.to_csv(os.path.join(DATA_DIR, 'Cyber_Document_Library.csv'), index=False)
cyber_doc_df.head()
```

```python
# 2) Equipment_Issues
n_rows = 10
equipment_issues_columns = ['Mission_Title', 'Issue_ID', 'Equipment_Type', 'Reported_By', 'Date_Reported', 'Status', 'Equipment Issues']
equipment_issues_df = pd.DataFrame({
    'Mission_Title': mission_titles,
    'Issue_ID': [f'ISSUE{i+1:03d}' for i in range(n_rows)],
    'Equipment_Type': np.random.choice(['Laptop', 'Router', 'Server', 'Switch'], n_rows),
    'Reported_By': np.random.choice(['User A', 'User B', 'User C'], n_rows),
    'Date_Reported': pd.date_range(datetime.today(), periods=n_rows).strftime('%Y-%m-%d'),
    'Status': np.random.choice(['Open', 'Closed', 'In Progress'], n_rows),
    'Equipment Issues': [random_text(100) for _ in range(n_rows)]
})
equipment_issues_df.to_csv(os.path.join(DATA_DIR, 'Equipment_Issues.csv'), index=False)
equipment_issues_df.head()
```

```python
# 3) FFIRs, Friction Points, Recommendations
n_rows = 10
ffirs_columns = ['Mission_Title', 'FFIR', 'Friction_Point', 'Recommendation', 'Observation', 'Action_Item', 'Summary']
ffirs_df = pd.DataFrame({
    'Mission_Title': mission_titles,
    'FFIR': [random_text(80) for _ in range(n_rows)],
    'Friction_Point': [random_text(80) for _ in range(n_rows)],
    'Recommendation': [random_text(80) for _ in range(n_rows)],
    'Observation': [random_text(80) for _ in range(n_rows)],
    'Action_Item': [random_text(80) for _ in range(n_rows)],
    'Summary': [random_text(80) for _ in range(n_rows)]
})
ffirs_df.to_csv(os.path.join(DATA_DIR, 'FFIRs_Friction_Points_Recommendations.csv'), index=False)
ffirs_df.head()
```

```python
# 4) Hardening Recommendations
n_rows = 10
hardening_columns = ['Mission_Title', 'Recommendation_ID', 'System_Affected', 'Priority', 'Status', 'Hardening Recommendation #231']
hardening_df = pd.DataFrame({
    'Mission_Title': mission_titles,
    'Recommendation_ID': [f'HR{i+1:03d}' for i in range(n_rows)],
    'System_Affected': np.random.choice(['System A', 'System B', 'System C'], n_rows),
    'Priority': np.random.choice(['High', 'Medium', 'Low'], n_rows),
    'Status': np.random.choice(['Implemented', 'Pending', 'Rejected'], n_rows),
    'Hardening Recommendation #231': [random_text(120) for _ in range(n_rows)]
})
hardening_df.to_csv(os.path.join(DATA_DIR, 'Hardening_Recommendations.csv'), index=False)
hardening_df.head()
```

```python
# 5) Mission Details
n_rows = 10
mission_details_columns = [
    'CY', 'Mission_Title', 'Terrain_System_Past_End_of_Life', 'Terrain_Air_Gapped', 'Intel_Driven_Hunt',
    'TMP_Signing_Delay', 'ATC_Signing_Delay', 'ROC_Drill_Completed', 'ROC_Drill_Date', 'OIV_Conducted',
    'OIV_Date', 'CTE_Utilized', 'CTE_Date', 'Range_Utilized', 'Equipment_Issues', 'Equipment_Shipping',
    'CED_Orders_Issues', 'CED_Orders_Issues_Details', 'Badge_Building_Issues', 'Account_Issues',
    'Account_Issue_Details', 'Equipment_Issues_Details'
]
mission_details_df = pd.DataFrame({
    'CY': np.random.choice(['2024', '2025'], n_rows),
    'Mission_Title': mission_titles,
    'Terrain_System_Past_End_of_Life': np.random.choice([True, False], n_rows),
    'Terrain_Air_Gapped': np.random.choice([True, False], n_rows),
    'Intel_Driven_Hunt': np.random.choice([True, False], n_rows),
    'TMP_Signing_Delay': np.random.choice([True, False], n_rows),
    'ATC_Signing_Delay': np.random.choice([True, False], n_rows),
    'ROC_Drill_Completed': np.random.choice([True, False], n_rows),
    'ROC_Drill_Date': pd.date_range(datetime.today(), periods=n_rows).strftime('%Y-%m-%d'),
    'OIV_Conducted': np.random.choice([True, False], n_rows),
    'OIV_Date': pd.date_range(datetime.today(), periods=n_rows).strftime('%Y-%m-%d'),
    'CTE_Utilized': np.random.choice([True, False], n_rows),
    'CTE_Date': pd.date_range(datetime.today(), periods=n_rows).strftime('%Y-%m-%d'),
    'Range_Utilized': np.random.choice([True, False], n_rows),
    'Equipment_Issues': np.random.choice([True, False], n_rows),
    'Equipment_Shipping': np.random.choice(['Shipped', 'Delayed', 'Lost'], n_rows),
    'CED_Orders_Issues': np.random.choice([True, False], n_rows),
    'CED_Orders_Issues_Details': np.random.choice(['None', 'Delay', 'Lost'], n_rows),
    'Badge_Building_Issues': np.random.choice(['None', 'Access Denied', 'Lost Badge'], n_rows),
    'Account_Issues': np.random.choice([True, False], n_rows),
    'Account_Issue_Details': np.random.choice(['None', 'Password Reset', 'Locked Out'], n_rows),
    'Equipment_Issues_Details': np.random.choice(['None', 'Damaged', 'Missing'], n_rows)
})
mission_details_df.to_csv(os.path.join(DATA_DIR, 'Mission_Details.csv'), index=False)
mission_details_df.head()
```

```python
# 6) Mission_Metrics
n_rows = 10
mission_metrics_columns = [
    'CY', 'Mission_Title', 'Number_of_JCO_Reports', 'Number_of_Tactical_Objectives', 'Number_of_Tactical_Tasks',
    'Number_of_Investigations', 'Number_of_Capability_Gaps', 'Number_of_Vulnerabilities',
    'Number_of_Hardening_Recommendations', 'Number_of_Terrain_TTPs', 'Number_of_ECOA_TTPs'
]
mission_metrics_df = pd.DataFrame({
    'CY': np.random.choice(['2024', '2025'], n_rows),
    'Mission_Title': mission_titles,
    'Number_of_JCO_Reports': np.random.randint(1, 10, n_rows),
    'Number_of_Tactical_Objectives': np.random.randint(1, 10, n_rows),
    'Number_of_Tactical_Tasks': np.random.randint(1, 10, n_rows),
    'Number_of_Investigations': np.random.randint(0, 5, n_rows),
    'Number_of_Capability_Gaps': np.random.randint(0, 5, n_rows),
    'Number_of_Vulnerabilities': np.random.randint(0, 5, n_rows),
    'Number_of_Hardening_Recommendations': np.random.randint(0, 5, n_rows),
    'Number_of_Terrain_TTPs': np.random.randint(0, 5, n_rows),
    'Number_of_ECOA_TTPs': np.random.randint(0, 5, n_rows)
})
mission_metrics_df.to_csv(os.path.join(DATA_DIR, 'Mission_Metrics.csv'), index=False)
mission_metrics_df.head()
```
